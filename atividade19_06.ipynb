```python
#4
numero = int(input("Digite um numero inteiro e positivo para calcular o fatorial: "))

if numero < 0:
    print("O fatorial é negativos digite um número inteiro e positivo.")
elif numero == 0:
    soma = 1 # Fatorial de 0 é 1
    print(f"O fatorial de {numero} é {soma}")
else:
    soma = 1
    for i in range(1, numero + 1):
        soma = soma * i
    print(f"O fatorial de {numero} é {soma}")
```

```python
# Solicita o valor do saque ao usuário
valor = int(input("Digite o valor do saque: "))

nota = 100

while valor > 0:

    qtdnotas = 0

    while valor >= nota:
        valor -= nota
        qtdnotas += 1

    if qtdnotas > 0:
        print(f"{qtdnotas} nota(s) de R${nota}")
    if nota == 100:
        nota = 50
    elif nota == 50:
        nota = 20
    elif nota == 20:
        nota = 10
    elif nota == 10:
        nota = 5
    elif nota == 5:
        nota = 2
    elif nota == 2:
        nota = 1
    elif nota == 1 and valor > 0:
        print(f"Não foi possível sacar completamente o valor. Restaram R${valor}")
        break
    elif nota == 1 and valor == 0:
        break
```
