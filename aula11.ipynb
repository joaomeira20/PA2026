```python
#entrada
nome = str(input("Digite seu nome: "))

b1 = float(input("digite a primeira nota do aluno: ").replace(',', '.'))
while b1 < 0 or b1 > 10:
  print ("nota invalida, digite uma nota de 0 a 10:")
  b1 = float(input("digite a primeira nota do aluno: ").replace(',', '.'))

b2 = float(input("digite a segunda nota do aluno: ").replace(',', '.'))
while b2 < 0 or b2 > 10:
  print ("nota invalida, digite uma nota de 0 a 10:")
  b2 = float(input("digite a segunda nota do aluno: ").replace(',', '.'))

b3 = float(input("digite a terceira nota do aluno: ").replace(',', '.'))
while b3 < 0 or b3 > 10:
  print ("nota invalida, digite uma nota de 0 a 10:")
  b3 = float(input("digite a terceira nota do aluno: ").replace(',', '.'))

b4 = float(input("digite a quarta nota do aluno: ").replace(',', '.'))
while b4 < 0 or b4 > 10:
  print ("nota invalida, digite uma nota de 0 a 10:")
  b4 = float(input("digite a quarta nota do aluno: ").replace(',', '.'))


#processamento
nota = (b1 + b2 + b3 + b4) / 4

#saida
if nota < 5:
  print (f"-O aluno {nome} foi Reprovado-")

elif nota >= 5 and nota < 7:
  print (f"-O aluno {nome} esta de Recuperação-")

else :
  print (f"-O aluno {nome} foi Aprovado-")
```
