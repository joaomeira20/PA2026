```python
# Escrever um jogo de adivinhação. O usuario tera 03 chances para adivinhar um numero de 0 a 10.
# o programa da uma mensagemse o palpite foi maior ou menor
# o programa avisa se esta quente ou frio

import random

# Generate a random number between 0 and 10
target_number = random.randint(0, 10)

chances = 3
guessed_correctly = False

print("Bem-vindo ao Jogo de Adivinhação!")
print("Estou pensando em um número entre 0 e 10. Você tem 3 chances para adivinhar.")

for attempt in range(1, chances + 1):
    try:
        guess_str = input(f"\nChance {attempt}/{chances}: Digite seu palpite (0-10): ")
        guess = int(guess_str)
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.\n")
        continue

    if not (0 <= guess <= 10):
        print("Seu palpite está fora do intervalo de 0 a 10. Por favor, tente novamente.\n")
        continue

    if guess == target_number:
        print(f"Parabéns! Você adivinhou o número {target_number} corretamente em {attempt} tentativas!")
        guessed_correctly = True
        break
    else:
        print("Não é isso!")
        # Provide higher/lower hint
        if guess < target_number:
            print("Seu palpite é muito baixo (menor).")
        else:
            print("Seu palpite é muito alto (maior).")

        # Provide hot/cold hint
        difference = abs(target_number - guess)
        if difference <= 2:
            print("Você está esquentando (quente)!")
        else:
            print("Você está frio (frio).")

if not guessed_correctly:
    print(f"\nFim de Jogo! Você ficou sem chances. O número em que eu estava pensando era {target_number}.")
```
