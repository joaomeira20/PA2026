```python
#3
numero1 = int(input("Digite o primeiro numero: "))
numero2 = int(input("Digite o segundo numero: "))

#Processamento e saida
if numero1 < numero2:
  print (f"o numero {numero1} e maior que o numero {numero2}.")

elif numero2 == numero1:
  print (f"os numeros {numero1} e {numero2} sao iguais.")

else:
  print (f'o numero {numero1} e menor que o numero {numero2}. ')
```

```python
#5
compra = float(input(f"digite o valor total da compra: "))
print (f"valor da compra: R$: {compra}")

if (compra >= 100):
    des = compra * 0.10
    total = compra - des
    print(f"O valor total da sua compra com 10% de desconto é de: {total}")

else:
    print(f"O valor total da sua compra é de: {compra}")
```

```python
#7
nome = str (input("digite seu nome: "))
idade = int (input("digiite sua idade: "))

#Processamento

if idade <= 12:
  print (f" {nome}, voce e uma criança")

elif idade  <= 17:
  print (f" {nome}, voce e um adolescente")

elif idade <= 59:
  print (f"{nome}, voce e um adulto")

else:
  print (f"{nome}, voce e um idoso")
```
